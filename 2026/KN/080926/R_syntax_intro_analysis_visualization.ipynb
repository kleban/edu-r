## 5. Порівняння та логічні оператори

Оператори порівняння повертають `TRUE` або `FALSE`:

- `==` — дорівнює;
- `!=` — не дорівнює;
- `>` — більше;
- `<` — менше;
- `>=` — більше або дорівнює;
- `<=` — менше або дорівнює.

Логічні оператори:

- `&` — логічне І;
- `|` — логічне АБО;
- `!` — заперечення.


````r
score <- 82

score >= 60
score == 100
score != 0

score >= 60 & score <= 100
score < 60 | score > 90
!(score < 60)
````

## 13. Пропущені значення `NA`

У реальних даних часто зустрічаються пропущені значення.

У R вони позначаються як:

```r
NA
```

Перевірити пропуски можна функцією `is.na()`.

Багато статистичних функцій мають параметр:


````r
scores <- c(70, 85, NA, 90, 65)

scores
is.na(scores)
sum(is.na(scores))

mean(scores)
mean(scores, na.rm = TRUE)
````

## 14. Робота з текстом

## 15. Власні функції

Функції дозволяють повторно використовувати код.

Загальний синтаксис:

```r
name <- function(arguments) {
  result <- ...
  return(result)
}
```

У R останній вираз функції також може повертатися автоматично.


````r
square <- function(x) {
  return(x ^ 2)
}

square(5)
square(10)

average_two <- function(a, b) {
  (a + b) / 2
}

average_two(10, 20)
````
